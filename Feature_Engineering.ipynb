1. What is a parameter?

A parameter is an internal variable of a machine learning model whose value is learned from the training data. Examples include the weights and biases in linear regression or neural networks.

2. What is correlation?

Correlation measures the statistical relationship between two variables, indicating how they change together. It ranges from -1 to +1. A value of +1 represents a perfect positive correlation, -1 represents a perfect negative correlation, and 0 represents no correlation.

3. What does negative correlation mean?

A negative correlation means that as one variable increases, the other variable tends to decrease. For example, as the temperature increases, the demand for heaters typically decreases.

4. Define Machine Learning. What are the main components in Machine Learning?

Machine Learning (ML) is a field of artificial intelligence that allows systems to learn from data and make predictions or decisions without being explicitly programmed. The main components typically include:
*   **Data:** The input information used for training and evaluating the model.
*   **Model:** The algorithm or structure that learns patterns from the data.
*   **Loss function:** A measure that quantifies how well the model's predictions match the actual values.
*   **Optimizer:** An algorithm used to adjust the model's parameters to minimize the loss function.
*   **Evaluation metric:** A measure used to assess the overall performance of the trained model.

5. How does loss value help in determining whether the model is good or not?

The loss value provides an indication of the difference between the model's predicted outputs and the actual outputs. A low loss value suggests that the model's predictions are close to the true values, indicating good performance. A high loss value suggests poor performance. The objective during training is to minimize this loss.

6. What are continuous and categorical variables?

**Continuous variables** are numerical variables that can take on any value within a given range (e.g., height, weight, temperature, price).
**Categorical variables** are variables that represent distinct categories or labels, with a limited number of possible values (e.g., gender, color, city, type of animal).

7. How do we handle categorical variables in Machine Learning? What are the common techniques?

Since machine learning models primarily work with numerical data, categorical variables need to be converted into a numerical format. Common techniques include:
*   **Label Encoding:** Assigning a unique integer to each category.
*   **One-Hot Encoding:** Creating new binary columns for each category, where a '1' indicates the presence of that category and a '0' indicates its absence.
*   **Ordinal Encoding:** Assigning integer values based on the order or rank of the categories, suitable for categorical variables with an inherent order.

8. What do you mean by training and testing a dataset?

**Training a dataset** is the process of using a portion of the available data to train a machine learning model. The model learns patterns and relationships from this data.
**Testing a dataset** is the process of evaluating the trained model's performance on a separate portion of the data that it has not seen before. This helps assess how well the model generalizes to new, unseen data.

9. What is sklearn.preprocessing?

`sklearn.preprocessing` is a module within the scikit-learn library in Python. It provides a wide range of tools and functions for preprocessing data before applying machine learning algorithms. This includes techniques for scaling, encoding, normalization, and imputation.

10. What is a Test set?

The test set is a subset of the original dataset that is reserved and not used during the training phase of a machine learning model. Its purpose is to provide an unbiased evaluation of the final model's performance on completely new data.

11. How do we split data for model fitting (training and testing) in Python?

In Python, a common way to split data into training and testing sets is by using the `train_test_split` function from the `sklearn.model_selection` module. This function randomly divides the data into specified proportions for training and testing.

12. How do you approach a Machine Learning problem?

A typical approach to solving a Machine Learning problem involves several steps:
*   **Problem Definition:** Clearly understanding the problem, the objective, and the desired outcome.
*   **Data Collection:** Gathering relevant data.
*   **Exploratory Data Analysis (EDA):** Analyzing and visualizing the data to understand its characteristics, identify patterns, and detect issues like missing values or outliers.
*   **Data Preprocessing:** Cleaning and preparing the data for modeling, including handling missing values, encoding categorical variables, and scaling features.
*   **Model Selection:** Choosing an appropriate machine learning algorithm based on the problem type and data characteristics.
*   **Model Training:** Training the selected model using the training data.
*   **Model Evaluation:** Assessing the trained model's performance using appropriate metrics on the test data.
*   **Hyperparameter Tuning:** Adjusting the model's hyperparameters to improve performance.
*   **Deployment (Optional):** Putting the trained model into production to make predictions on new data.

13. Why do we have to perform EDA before fitting a model to the data?

Performing EDA before fitting a model is essential because it helps to:
*   Gain a deep understanding of the data's structure, distributions, and relationships between variables.
*   Identify potential issues in the data, such as missing values, outliers, or inconsistencies, which need to be addressed during preprocessing.
*   Discover patterns and insights that can guide feature engineering and the selection of the most suitable model.
*   Visualize the data to communicate findings and understand its characteristics better.

17. What is causation? Explain difference between correlation and causation with an example.

**Causation** means that one event or variable directly causes another event or variable to occur. There is a cause-and-effect relationship.
**Correlation** indicates that two variables are related or tend to change together, but it does not necessarily mean that one causes the other.
**Example:** There is often a positive correlation between ice cream sales and crime rates (both tend to increase during the summer months). However, ice cream sales do not cause crime. The underlying cause for the increase in both is likely the warmer weather, which leads to more people being outside and potentially more opportunities for both activities.

18. What is an Optimizer? What are different types of optimizers? Explain each with an example.

An optimizer is an algorithm used during the training of a machine learning model to adjust the model's parameters (weights and biases) in order to minimize the loss function. Different types of optimizers exist:
*   **Gradient Descent:** A basic optimization algorithm that updates parameters in the direction opposite to the gradient of the loss function, moving towards the minimum.
*   **Stochastic Gradient Descent (SGD):** A variation of gradient descent that calculates the gradient and updates parameters using only a single training example or a small batch of examples at each iteration. This makes it faster for large datasets.
*   **Adam:** An adaptive learning rate optimization algorithm that uses estimations of the first and second moments of the gradients to adjust the learning rate for each parameter individually. It is widely used due to its effectiveness.

20. What does model.fit() do? What arguments must be given?

The `model.fit()` method is used to train a machine learning model. It fits the model to the provided training data. The essential arguments typically required are:
*   The training data features (usually denoted as `X_train`), which is a matrix or DataFrame containing the independent variables.
*   The training data target variable (usually denoted as `y_train`), which is a vector or Series containing the dependent variable.

21. What does model.predict() do? What arguments must be given?

The `model.predict()` method is used after a model has been trained using `model.fit()`. It takes new, unseen data as input and uses the learned patterns from training to make predictions on this data. The essential argument is:
*   The data for which you want to make predictions (usually denoted as `X_test`), which should have the same feature structure as the training data.

23. What is feature scaling? How does it help in Machine Learning?

Feature scaling is a data preprocessing technique used to standardize the range of independent variables or features in the dataset. It transforms the values of features so that they have a similar scale.
It helps in Machine Learning by preventing features with larger numerical ranges from disproportionately influencing the model, especially for algorithms that are sensitive to the scale of the input features (e.g., gradient descent based algorithms, k-Nearest Neighbors, Support Vector Machines).

24. How do we perform scaling in Python?

In Python, you can perform feature scaling using classes from the `sklearn.preprocessing` module. Common scalers include `StandardScaler`, which scales features to have zero mean and unit variance, and `MinMaxScaler`, which scales features to a specified range (usually between 0 and 1).
