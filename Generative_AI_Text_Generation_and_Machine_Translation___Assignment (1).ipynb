**Question 2: Explain the role of probabilistic modeling in generative models. How do these models differ from discriminative models?**


Answer: Probabilistic modeling is the core of generative models. It enables the model to learn the underlying probability distribution of the training data instead of simply predicting outputs. A generative model learns the joint probability distribution, P(X,Y), where X represents the input data and Y represents the corresponding labels or classes. By learning this distribution, the model understands how the data is generated.

This allows generative models to:

Generate new data samples similar to the training data.
Estimate the likelihood of observed data.
Handle missing or incomplete data effectively.
Perform classification using Bayes' theorem by computing the probability of a class given the input.

Examples of generative models include Naive Bayes, Gaussian Mixture Models (GMMs), Hidden Markov Models (HMMs), Variational Autoencoders (VAEs), and Generative Adversarial Networks (GANs).

Difference Between Generative and Discriminative Models

Generative models learn the joint probability distribution P(X,Y), which helps them understand how the data is generated. As a result, they can generate new data samples as well as perform classification.

In contrast, discriminative models learn the conditional probability distribution P(Y∣X). Their main objective is to predict the correct output or class label for a given input by learning the decision boundary between different classes. They do not generate new data.

Generative models are commonly used for tasks such as image generation, text generation, anomaly detection, and data synthesis. Discriminative models are mainly used for classification and regression tasks, such as spam detection, sentiment analysis, and image classification.

Examples of discriminative models include Logistic Regression, Support Vector Machines (SVMs), Decision Trees, Random Forests, and Neural Networks.

**Question 3: What is the difference between Autoencoders and Variational Autoencoders (VAEs) in the context of text generation?**




Answer 3: Difference between Autoencoders and Variational Autoencoders (VAEs) in Text Generation

Both Autoencoders (AEs) and Variational Autoencoders (VAEs) are neural network architectures used for dimensionality reduction and learning latent representations of data. However, their fundamental design and the nature of their latent space differ significantly, especially concerning their application in text generation.

### Autoencoders (AEs)

An Autoencoder is a type of neural network that learns an efficient data encoding (a compressed representation) in an unsupervised manner. It consists of two main parts:

1.  **Encoder:** Maps the input data `X` to a lower-dimensional latent space representation `z`. The encoder learns `q(z|X)`.
2.  **Decoder:** Maps the latent space representation `z` back to a reconstruction of the original input `X'`. The decoder learns `p(X'|z)`.

**In the context of text generation:**
*   **Encoding:** An AE for text would typically take an input text sequence (e.g., token IDs, word embeddings) and compress it into a fixed-size vector `z` (the latent representation).
*   **Decoding:** The decoder would then try to reconstruct the original text sequence from `z`.
*   **Generation:** A pure AE is **not inherently a generative model** in the sense of creating *new, diverse* samples. Its latent space `z` is typically not structured in a way that allows for meaningful sampling. If you sample a random `z` from its latent space, there's no guarantee that the decoder will produce a coherent or realistic text. The latent space of a standard AE can have 'holes' or discontinuities, meaning small changes in `z` might lead to wildly different or nonsensical decoded outputs.

### Variational Autoencoders (VAEs)

VAEs are a type of generative model that extend AEs by introducing a probabilistic approach to the encoder and a regularization technique to the latent space. This probabilistic nature allows VAEs to generate new data points by sampling from the learned latent distribution.

1.  **Probabilistic Encoder:** Instead of mapping the input `X` to a single point `z` in the latent space, the encoder maps `X` to the parameters of a probability distribution (typically a Gaussian distribution) in the latent space. Specifically, it outputs a `mean (μ)` and a `log-variance (log σ²)` for each dimension of `z`.
2.  **Reparameterization Trick:** To allow gradients to flow through the sampling process during training, a "reparameterization trick" is used: `z = μ + σ * ε`, where `ε` is a random sample from a standard normal distribution (N(0, 1)). This makes the latent space continuous and smooth.
3.  **Regularized Latent Space:** VAEs add a **Kullback-Leibler (KL) divergence** term to their loss function. This term forces the learned latent distributions (`q(z|X)`) to be close to a prior distribution (typically a standard normal distribution `p(z)`). This regularization ensures that the latent space is continuous and well-structured, making it easy to sample from.
4.  **Decoder:** Similar to AEs, the decoder takes a latent sample `z` and reconstructs the input `X'`.

**In the context of text generation:**
*   **Encoding:** The encoder compresses text into a probabilistic representation in the latent space, providing a `mean` and `log-variance` for each text snippet.
*   **Decoding:** The decoder reconstructs the text from a sampled latent vector `z`.
*   **Generation:** This is where VAEs shine. Because their latent space is regularized to be continuous and usually follows a simple distribution (like a standard normal), you can sample a random `z` directly from this prior distribution. Feeding this `z` to the decoder will then produce a *new, coherent, and diverse* text sequence. This capability makes VAEs powerful for tasks like generating novel sentences, conditional text generation, or even controlling text attributes by manipulating the latent space.

### Key Differences Summarized:

| Feature              | Autoencoder (AE)                                      | Variational Autoencoder (VAE)                                    |
| :------------------- | :---------------------------------------------------- | :--------------------------------------------------------------- |
| **Latent Space**     | Deterministic point `z` for each input.               | Probabilistic distribution (`μ`, `log σ²`) for each input.       |
| **Generative Cap.**  | Poor. Random samples from `z` often yield nonsense.   | Excellent. Smooth, continuous latent space allows meaningful sampling for generation. |
| **Loss Function**    | Reconstruction Loss (e.g., MSE, Cross-entropy).       | Reconstruction Loss + KL Divergence (regularization).            |
| **Latent Space Struct.**| Can be unstructured, leading to 'holes'.             | Explicitly regularized to be continuous and follow a prior distribution (e.g., Gaussian). |
| **Primary Use**      | Dimensionality reduction, feature learning, denoising. | Generative modeling, sampling, text synthesis, disentangled representations. |

In essence, while AEs learn a compressed representation, VAEs learn a *generative distribution* of that compressed representation, making them far more suitable for tasks requiring the creation of novel and diverse text content.

**Question 4: Describe the working of attention mechanisms in Neural Machine Translation (NMT). Why are they critical?**

Answer to Question 4: Attention Mechanisms in Neural Machine Translation (NMT)

### Working of Attention Mechanisms in NMT

**Neural Machine Translation (NMT)** models, particularly those based on the **encoder-decoder architecture**, were historically limited by their inability to effectively handle long sentences. The encoder would compress the entire source sentence into a single fixed-size context vector, leading to information loss, especially for longer inputs. This bottleneck severely impacted translation quality.

**Attention mechanisms** were introduced to address this limitation by allowing the decoder to "attend" to different parts of the source sentence at each step of generating the target sentence. Instead of relying on a single context vector, attention creates a dynamic context vector that is a weighted sum of the encoder's hidden states (or outputs).

Here's how it generally works:

1.  **Encoder Processing:** The encoder (often an RNN like LSTM or GRU, or more recently, a Transformer encoder) processes the source sentence word by word and generates a sequence of hidden states. Each hidden state ideally encapsulates information about the word at that position and its context in the source sentence.

2.  **Decoder Querying:** At each step of generating a target word, the decoder (another RNN or Transformer decoder) computes a query based on its current hidden state.

3.  **Alignment Score Calculation:** This query is then compared to all of the encoder's hidden states. A scoring function (e.g., dot product, concatenation followed by a feed-forward network) calculates an "alignment score" or "energy" for each source hidden state with respect to the current decoder state. This score indicates how relevant each source word is to generating the current target word.

4.  **Attention Weights:** These alignment scores are typically normalized using a softmax function to produce "attention weights." These weights sum to 1 and represent a probability distribution over the source words. A higher weight means that the corresponding source word is more important for the current decoding step.

5.  **Context Vector Creation:** The attention weights are then used to compute a weighted sum of the encoder's hidden states. This weighted sum forms the **context vector**. This context vector dynamically highlights the most relevant parts of the source sentence for generating the current target word.

6.  **Decoder Prediction:** The context vector is then concatenated with the decoder's current hidden state (or fed into the decoder in other ways) to help predict the next word in the target sequence.

This process allows the decoder to focus its "attention" on different parts of the input sequence as it translates, mimicking how a human translator might look back at specific words in the source sentence while translating.

### Why Attention Mechanisms are Critical

Attention mechanisms are critical for NMT (and many other sequence-to-sequence tasks) for several key reasons:

1.  **Resolving the Fixed-Size Context Bottleneck:** This was the primary motivation. Attention eliminates the need to compress all information from a potentially long source sentence into a single fixed-length vector. Instead, the decoder can access all encoder hidden states directly, drawing relevant information as needed.

2.  **Handling Long-Range Dependencies:** By allowing direct access to past encoder states, attention helps models capture long-range dependencies between source and target words more effectively than traditional RNNs alone. This significantly improves translation quality for longer sentences.

3.  **Improved Performance for Long Sentences:** Before attention, the quality of translation would degrade significantly for sentences beyond a certain length. Attention mechanisms dramatically improved NMT performance, making it practical for real-world applications with varying sentence lengths.

4.  **Interpretability/Alignment:** Attention weights provide a form of soft alignment between source and target words. By visualizing these weights, we can see which source words the model focused on when generating each target word. This offers valuable insights into the model's decision-making process, making it more interpretable.

5.  **Foundation for Transformer Architecture:** Attention mechanisms, particularly the self-attention mechanism, are the sole building block of the highly successful Transformer architecture. Transformers, which eschew recurrent connections entirely, leverage attention to process all input tokens in parallel, further enhancing efficiency and performance for NMT and other tasks.

6.  **Enhanced Gradient Flow:** In deep RNNs, gradients can vanish or explode over long sequences. Attention helps alleviate this by creating direct shortcuts between the decoder and relevant parts of the encoder output, facilitating better gradient flow during training.

In essence, attention mechanisms transformed NMT from a research curiosity into a powerful, commercially viable technology by enabling models to focus selectively on relevant information, overcoming the limitations of fixed-context representations, and providing a degree of interpretability.


**Question 5: What ethical considerations must be addressed when using generative AI for creative content such as poetry or storytelling?**



Answer 5: Ethical Considerations for Generative AI in Creative Content

When using generative AI for creative content like poetry or storytelling, several critical ethical considerations must be addressed to ensure responsible and fair development and deployment:

1.  **Authorship and Originality:**
    *   **Attribution:** Who is the author—the human who prompted the AI, the AI itself, or the developers of the AI? Clear guidelines are needed for attribution, especially in commercial contexts.
    *   **Originality:** Does AI-generated content truly possess originality, or is it merely a sophisticated recombination of its training data? This impacts copyright, intellectual property, and artistic value.
    *   **Plagiarism:** Generative AI models can inadvertently reproduce elements from their training data. Ensuring that generated content does not plagiarize existing works is crucial.

2.  **Bias and Stereotypes:**
    *   **Reinforcement of Bias:** AI models learn from vast datasets, which often reflect societal biases. If not carefully curated, AI can perpetuate or amplify harmful stereotypes (e.g., gender roles, racial characteristics) in generated stories or character descriptions.
    *   **Lack of Diversity:** Over-reliance on homogenous training data can lead to a lack of diversity in generated narratives and characters, failing to represent a broad range of human experiences and perspectives.

3.  **Misinformation and Malicious Use:**
    *   **Deepfakes and Fabricated Narratives:** AI can generate highly convincing, yet entirely fictitious, stories or poems. This capability could be abused to create propaganda, spread misinformation, or impersonate individuals, leading to reputational damage or public deception.
    *   **Exploitation of Vulnerabilities:** Malicious actors could use AI to generate content that exploits emotional or psychological vulnerabilities, for example, by crafting manipulative narratives.

4.  **Economic Impact on Human Creators:**
    *   **Displacement of Jobs:** The rise of AI-generated content raises concerns about job displacement for writers, poets, artists, and other creative professionals.
    *   **Fair Compensation:** If AI-generated content is used commercially, how will human contributors (data providers, prompt engineers, editors) be compensated? What about the original creators whose works formed the training data?

5.  **Transparency and Disclosure:**
    *   **Identifying AI-Generated Content:** Should there be a requirement to disclose when content has been generated by AI? This helps consumers distinguish between human and machine creativity and informs their judgment about the content's authenticity and intent.
    *   **Explainability:** Understanding *how* the AI arrived at a particular creative output can be challenging. Lack of explainability can hinder efforts to mitigate bias or identify sources of problematic content.

6.  **Ethical Responsibility and Accountability:**
    *   **Who is Accountable?** If AI-generated content causes harm (e.g., defamation, hate speech), who is legally and ethically responsible: the developer, the user, or the AI itself?
    *   **Value Alignment:** Ensuring that the AI's creative outputs align with human ethical values and societal norms is an ongoing challenge that requires continuous oversight and refinement.

Addressing these ethical considerations requires a multidisciplinary approach involving AI researchers, ethicists, policymakers, legal experts, and the creative community to develop robust frameworks, guidelines, and technological solutions.

**Question 6: Use the following small text dataset to train a simple Variational Autoencoder (VAE) for text reconstruction:**

["The sky is blue", "The sun is bright", "The grass is green",
"The night is dark", "The stars are shining"]

1. Preprocess the data (tokenize and pad the sequences).
2. Build a basic VAE model for text reconstruction.
3. Train the model and show how it reconstructs or generates similar sentences.

**Include your code, explanation, and sample outputs.**

````python
# Answer 6: Text Reconstruction with a Variational Autoencoder (VAE)

import numpy as np
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Embedding, LSTM, Dense, Lambda
from tensorflow.keras import backend as K
import matplotlib.pyplot as plt

# --- 1. Data Preprocessing ---
# Provided text dataset
dataset = [
    "The sky is blue",
    "The sun is bright",
    "The grass is green",
    "The night is dark",
    "The stars are shining"
]

# Tokenize and build vocabulary
tokenizer = Tokenizer(num_words=None, oov_token='<unk>', filters='')
tokenizer.fit_on_texts(dataset)

vocabulary_size = len(tokenizer.word_index) + 1 # +1 for padding/OOV token

# Convert text to sequences of integers
sequences = tokenizer.texts_to_sequences(dataset)

# Pad sequences to a fixed length (MAX_LEN)
max_sequence_length = max([len(seq) for seq in sequences])
MAX_LEN = max_sequence_length # Using actual max length, not adding extra tokens explicitly here for simplicity

padded_sequences = pad_sequences(sequences, maxlen=MAX_LEN, padding='post', value=0) # 0 for padding

# Create word to index and index to word mappings
word_to_idx = tokenizer.word_index
idx_to_word = {v: k for k, v in word_to_idx.items()}
idx_to_word[0] = '<pad>'

def sequence_to_text(sequence):
    return ' '.join([idx_to_word.get(idx, '<unk>') for idx in sequence if idx != 0])

print(f"Padded sequences:\n{padded_sequences}")
print(f"Example reconstruction: {sequence_to_text(padded_sequences[0])}\n")

# --- 2. Build a Basic VAE Model for Text Reconstruction ---
# VAE parameters
LATENT_DIM = 64
EMBEDDING_DIM = 100
LSTM_UNITS = 128

# Encoder
encoder_inputs = Input(shape=(MAX_LEN,))
encoder_embedding = Embedding(vocabulary_size, EMBEDDING_DIM)(encoder_inputs)
encoder_lstm = LSTM(LSTM_UNITS, return_state=True)(encoder_embedding)
_, state_h, state_c = encoder_lstm # Only need states for VAE
encoder_states = [state_h, state_c] # Using h and c for LSTM state

z_mean = Dense(LATENT_DIM, name='z_mean')(state_h)
z_log_var = Dense(LATENT_DIM, name='z_log_var')(state_h)

# Sampling function for the latent space
def sampling(args):
    z_mean, z_log_var = args
    epsilon = K.random_normal(shape=(K.shape(z_mean)[0], LATENT_DIM))
    return z_mean + K.exp(0.5 * z_log_var) * epsilon

z = Lambda(sampling, output_shape=(LATENT_DIM,), name='z_sampled')([z_mean, z_log_var])

# Decoder
decoder_inputs = Input(shape=(MAX_LEN,))
decoder_embedding = Embedding(vocabulary_size, EMBEDDING_DIM)(decoder_inputs)
decoder_lstm = LSTM(LSTM_UNITS, return_sequences=True, return_state=True)(decoder_embedding, initial_state=encoder_states) # Using encoder states as initial state for decoder
decoder_outputs = Dense(vocabulary_size, activation='softmax')(decoder_lstm[0])

# VAE Model
vae = Model([encoder_inputs, decoder_inputs], decoder_outputs)

# Encoder for inference (to get z_mean, z_log_var)
encoder_model = Model(encoder_inputs, [z_mean, z_log_var])

# Decoder for generation (given a latent vector and a dummy input sequence)
decoder_latent_input = Input(shape=(LATENT_DIM,))

# Map latent vector to initial LSTM state
latent_to_state_h = Dense(LSTM_UNITS, activation='tanh')(decoder_latent_input)
latent_to_state_c = Dense(LSTM_UNITS, activation='tanh')(decoder_latent_input)
decoder_initial_states_from_latent = [latent_to_state_h, latent_to_state_c]

# Reuse the decoder_embedding layer (which takes decoder_inputs) and the LSTM logic
# The `decoder_inputs` will be the placeholder sequence for generation.
decoder_outputs_gen, state_h_gen, state_c_gen = LSTM(LSTM_UNITS, return_sequences=True, return_state=True)(decoder_embedding, initial_state=decoder_initial_states_from_latent)
decoder_outputs_gen = Dense(vocabulary_size, activation='softmax')(decoder_outputs_gen)

# Corrected: decoder_states_inputs were unused and removed from the model inputs/outputs
decoder_model = Model([decoder_latent_input, decoder_inputs], [decoder_outputs_gen, state_h_gen, state_c_gen])

vae.summary()

# --- 3. VAE Loss Function ---
def vae_loss(y_true, y_pred):
    # Reconstruction loss (sparse categorical cross-entropy for sequence prediction)
    # We need to reshape y_true and y_pred for `sparse_categorical_crossentropy`
    y_true_flat = K.flatten(y_true)
    y_pred_reshaped = K.reshape(y_pred, (-1, K.shape(y_pred)[-1]))

    # Only calculate loss for non-padding tokens
    mask = tf.cast(tf.not_equal(y_true_flat, 0), dtype=tf.float32)
    reconstruction_loss = K.sparse_categorical_crossentropy(y_true_flat, y_pred_reshaped) * mask
    reconstruction_loss = K.sum(reconstruction_loss)

    # KL divergence loss
    kl_loss = -0.5 * K.sum(1 + z_log_var - K.square(z_mean) - K.exp(z_log_var), axis=-1)
    kl_loss = K.sum(kl_loss) # Sum over batch elements as well

    return reconstruction_loss + kl_loss

vae.add_loss(vae_loss(encoder_inputs, vae.output)) # Add VAE loss as a model loss
vae.compile(optimizer='adam')

# Prepare target data for reconstruction (input is also the target for reconstruction)
X_train = padded_sequences
Y_train = padded_sequences

print("Input data shape (X_train):", X_train.shape)
print("Target data shape (Y_train):", Y_train.shape)

# --- 4. Train the VAE Model ---
print("\nStarting VAE training...")
history = vae.fit([X_train, X_train], None, epochs=50, batch_size=len(dataset), verbose=0) # None for y_true as loss is custom

plt.figure(figsize=(8, 5))
plt.plot(history.history['loss'])
plt.title('VAE Training Loss')
plt.ylabel('Loss')
plt.xlabel('Epoch')
plt.show()

# --- 5. Reconstruction and Generation ---

print("\n--- Reconstruction Examples ---")
for i, original_sequence_padded in enumerate(padded_sequences):
    original_sequence_expanded = np.expand_dims(original_sequence_padded, axis=0)

    # The VAE model is trained to reconstruct its input, so feed the same sequence to both inputs.
    reconstructed_output = vae.predict([original_sequence_expanded, original_sequence_expanded])

    # Get the predicted sequence by taking argmax (most probable word) for each time step
    # We only care about the first batch item since we predict one by one
    reconstructed_sequence_indices = np.argmax(reconstructed_output[0], axis=-1)

    original_text = sequence_to_text(original_sequence_padded)
    reconstructed_text = sequence_to_text(reconstructed_sequence_indices)

    print(f"Original: {original_text}")
    print(f"Reconstructed: {reconstructed_text}")
    print("-" * 30)

# --- Sentence Generation ---
print("\n--- Generation Examples (from latent space samples) ---")


num_generations = 3
for _ in range(num_generations):
    # Sample a random point from the latent space (e.g., from a standard normal distribution)
    random_latent_vector = np.random.normal(size=(1, LATENT_DIM))

    # Initialize decoder input sequence with padding or an explicit start token if used
    dummy_decoder_input_for_generation = np.zeros((1, MAX_LEN)) # Start with all padding

    # The decoder_model now takes the latent vector and the dummy sequence to generate the full output
    predicted_output_seq_probs, _, _ = decoder_model.predict([random_latent_vector, dummy_decoder_input_for_generation])
    generated_sentence_indices = np.argmax(predicted_output_seq_probs[0], axis=-1)

    generated_text = sequence_to_text(generated_sentence_indices)
    print(f"Generated: {generated_text}")
````

**Question 8: Implement a simple attention-based encoder-decoder model for English-to-Spanish translation using Tensorflow or PyTorch.**


````python
# Answer 8: Attention-based Encoder-Decoder Model for English-to-Spanish Translation (TensorFlow)

import tensorflow as tf
import numpy as np
import unicodedata
import re
import time

# --- 1. Data Preparation ---
# Converts the unicode file to ascii
def unicode_to_ascii(s):
    return ''.join(c for c in unicodedata.normalize('NFD', s) if unicodedata.category(c) != 'Mn')

def preprocess_sentence(w):
    w = unicode_to_ascii(w.lower().strip())
    w = re.sub(r"([?.!,¿])", r" \1 ", w)
    w = re.sub(r'["()]', '', w)
    w = re.sub(r"[^a-zA-Z?.!,¿]+", " ", w)
    w = w.strip()
    w = '<start> ' + w + ' <end>'
    return w

en_sentences = [
    "I am a student.",
    "How are you?",
    "Where is the library?",
    "Please help me.",
    "The weather is nice today.",
    "I love to learn deep learning.",
    "This is a simple sentence."
]

es_sentences = [
    "Soy un estudiante.",
    "¿Cómo estás?",
    "¿Dónde está la biblioteca?",
    "Por favor, ayúdame.",
    "El clima es agradable hoy.",
    "Me encanta aprender aprendizaje profundo.",
    "Esta es una frase sencilla."
]

processed_en_sentences = [preprocess_sentence(s) for s in en_sentences]
processed_es_sentences = [preprocess_sentence(s) for s in es_sentences]

# --- 2. Tokenization and Vocabulary ---
def tokenize_and_pad(lang_sentences):
    tokenizer = tf.keras.preprocessing.text.Tokenizer(filters='', oov_token='<unk>')
    tokenizer.fit_on_texts(lang_sentences)
    tensor = tokenizer.texts_to_sequences(lang_sentences)
    tensor = tf.keras.preprocessing.sequence.pad_sequences(tensor, padding='post')
    return tensor, tokenizer

input_tensor, inp_lang_tokenizer = tokenize_and_pad(processed_en_sentences)
target_tensor, targ_lang_tokenizer = tokenize_and_pad(processed_es_sentences)

max_length_targ = max(len(t) for t in target_tensor)
max_length_inp = max(len(t) for t in input_tensor)

# --- 3. Create tf.data.Dataset ---
BUFFER_SIZE = len(input_tensor)
BATCH_SIZE = 2
embedding_dim = 256
units = 1024
vocab_inp_size = len(inp_lang_tokenizer.word_index) + 1
vocab_targ_size = len(targ_lang_tokenizer.word_index) + 1

dataset = tf.data.Dataset.from_tensor_slices((input_tensor, target_tensor)).shuffle(BUFFER_SIZE)
dataset = dataset.batch(BATCH_SIZE, drop_remainder=True)

# --- 4. Encoder Architecture ---
class Encoder(tf.keras.Model):
    def __init__(self, vocab_size, embedding_dim, enc_units, batch_sz):
        super(Encoder, self).__init__()
        self.batch_sz = batch_sz
        self.enc_units = enc_units
        self.embedding = tf.keras.layers.Embedding(vocab_size, embedding_dim)
        self.gru = tf.keras.layers.GRU(self.enc_units,
                                       return_sequences=True,
                                       return_state=True,
                                       recurrent_initializer='glorot_uniform')

    def call(self, x, hidden):
        x = self.embedding(x)
        output, state = self.gru(x, initial_state=hidden)
        return output, state

    def initialize_hidden_state(self):
        return tf.zeros((self.batch_sz, self.enc_units))

# --- 5. Attention Mechanism (Bahdanau Attention) ---
class BahdanauAttention(tf.keras.layers.Layer):
    def __init__(self, units):
        super(BahdanauAttention, self).__init__()
        self.W1 = tf.keras.layers.Dense(units)
        self.W2 = tf.keras.layers.Dense(units)
        self.V = tf.keras.layers.Dense(1)

    def call(self, query, values):
        query_with_time_axis = tf.expand_dims(query, 1)
        score = self.V(tf.nn.tanh(
            self.W1(query_with_time_axis) + self.W2(values)))
        attention_weights = tf.nn.softmax(score, axis=1)
        context_vector = attention_weights * values
        context_vector = tf.reduce_sum(context_vector, axis=1)
        return context_vector, attention_weights

# --- 6. Decoder Architecture ---
class Decoder(tf.keras.Model):
    def __init__(self, vocab_size, embedding_dim, dec_units, batch_sz):
        super(Decoder, self).__init__()
        self.batch_sz = batch_sz
        self.dec_units = dec_units
        self.embedding = tf.keras.layers.Embedding(vocab_size, embedding_dim)
        self.gru = tf.keras.layers.GRU(self.dec_units,
                                       return_sequences=True,
                                       return_state=True,
                                       recurrent_initializer='glorot_uniform')
        self.fc = tf.keras.layers.Dense(vocab_size)
        self.attention = BahdanauAttention(self.dec_units)

    def call(self, x, hidden, enc_output):
        context_vector, attention_weights = self.attention(hidden, enc_output)
        x = self.embedding(x)
        x = tf.concat([tf.expand_dims(context_vector, 1), x], axis=-1)
        output, state = self.gru(x, initial_state=hidden)
        output = tf.reshape(output, (-1, output.shape[2]))
        x = self.fc(output)
        return x, state, attention_weights

# --- 7. Optimizer and Loss Function ---
encoder = Encoder(vocab_inp_size, embedding_dim, units, BATCH_SIZE)
decoder = Decoder(vocab_targ_size, embedding_dim, units, BATCH_SIZE)

optimizer = tf.keras.optimizers.Adam()
loss_object = tf.keras.losses.SparseCategoricalCrossentropy(
    from_logits=True, reduction='none')

def loss_function(real, pred):
    mask = tf.math.logical_not(tf.math.equal(real, 0))
    loss_ = loss_object(real, pred)
    mask = tf.cast(mask, dtype=loss_.dtype)
    loss_ *= mask
    return tf.reduce_mean(loss_)

# --- 8. Training Loop ---
@tf.function
def train_step(inp, targ, enc_hidden):
    loss = 0
    with tf.GradientTape() as tape:
        enc_output, enc_hidden = encoder(inp, enc_hidden)
        dec_hidden = enc_hidden
        dec_input = tf.expand_dims([targ_lang_tokenizer.word_index['<start>']] * BATCH_SIZE, 1)

        for t in range(1, targ.shape[1]):
            predictions, dec_hidden, _ = decoder(dec_input, dec_hidden, enc_output)
            loss += loss_function(targ[:, t], predictions)
            dec_input = tf.expand_dims(targ[:, t], 1)

    batch_loss = (loss / int(targ.shape[1]))
    variables = encoder.trainable_variables + decoder.trainable_variables
    gradients = tape.gradient(loss, variables)
    optimizer.apply_gradients(zip(gradients, variables))
    return batch_loss

EPOCHS = 10
print("\nStarting Training...")
for epoch in range(EPOCHS):
    start = time.time()
    enc_hidden = encoder.initialize_hidden_state()
    total_loss = 0

    for (batch, (inp, targ)) in enumerate(dataset.take(len(input_tensor)//BATCH_SIZE)):
        batch_loss = train_step(inp, targ, enc_hidden)
        total_loss += batch_loss
        if batch % 1 == 0:
            print('Epoch {} Batch {} Loss {:.4f}'.format(epoch + 1, batch, batch_loss.numpy()))
    print('Epoch {} Loss {:.4f}'.format(epoch + 1, total_loss / (len(input_tensor)//BATCH_SIZE)))
    print('Time taken for 1 epoch {} sec\n'.format(time.time() - start))

# --- 9. Translate Function (Inference) ---
def translate(sentence):
    sentence = preprocess_sentence(sentence)
    inputs = [inp_lang_tokenizer.word_index[i] for i in sentence.split(' ')]
    inputs = tf.keras.preprocessing.sequence.pad_sequences([inputs],
                                                           maxlen=max_length_inp,
                                                           padding='post')
    inputs = tf.convert_to_tensor(inputs)

    result = ''
    hidden = [tf.zeros((1, units))]
    enc_out, enc_hidden = encoder(inputs, hidden)

    dec_hidden = enc_hidden
    dec_input = tf.expand_dims([targ_lang_tokenizer.word_index['<start>']], 0)

    for t in range(max_length_targ):
        predictions, dec_hidden, attention_weights = decoder(dec_input,
                                                             dec_hidden,
                                                             enc_out)

        predicted_id = tf.argmax(predictions[0]).numpy()
        result += targ_lang_tokenizer.index_word[predicted_id] + ' '

        if targ_lang_tokenizer.index_word[predicted_id] == '<end>':
            return result, sentence

        dec_input = tf.expand_dims([predicted_id], 0)

    return result, sentence


print("\n--- Translation Examples ---")
result, sentence = translate("I am a student.")
print(f"Input: {sentence}")
print(f"Translated output: {result}")

result, sentence = translate("Where is the library?")
print(f"Input: {sentence}")
print(f"Translated output: {result}")

result, sentence = translate("I love to learn deep learning.")
print(f"Input: {sentence}")
print(f"Translated output: {result}")
````

**Question 10: Imagine you are building a creative writing assistant for a publishing company. The assistant should generate story plots and character descriptions using Generative AI. Describe how you would design the system, including model selection, training data, bias mitigation, and evaluation methods. Explain the real-world challenges you might face.**

Answer: System Design for a Creative Writing Assistant:

1.  **Model Selection:**
    *   For **Story Plots**: A large Language Model (LLM) such as GPT-3/GPT-4, or a fine-tuned transformer-based model (e.g., T5, LLaMA), would be ideal. These models excel at understanding narrative structures and generating coherent, long-form text. A hierarchical approach could involve one model for high-level plot points and another for detailed scene generation.
    *   For **Character Descriptions**: Similar LLMs, potentially fine-tuned on character archetypes, personality traits, and descriptive language. An additional layer might ensure consistency between character descriptions and the generated plot. Pre-trained models can be fine-tuned to align with the publishing house's specific stylistic requirements.

2.  **Training Data:**
    *   **Source**: A vast corpus of existing novels, short stories, screenplays, character encyclopedias, and literary analyses. Incorporating the publishing company's own archives would be crucial for stylistic consistency and brand voice.
    *   **Preprocessing**: Data must be cleaned (removing boilerplate, formatting issues), tokenized, and potentially tagged by genre, theme, character role, etc., to allow for more controlled and nuanced generation.
    *   **Diversity**: It is critical to include works from a wide array of authors, cultures, periods, and perspectives to prevent the generation of narrow, stereotypical, or repetitive outputs.

3.  **Bias Mitigation:**
    *   **Data Curation**: Actively select and balance training data to represent diverse characters, settings, and plotlines. Explicitly filter out or rebalance data known to contain harmful stereotypes (e.g., based on gender, race, sexuality, ability).
    *   **Model Fine-tuning**: Fine-tune the chosen LLMs with deliberately diverse and inclusive examples. Techniques like 'debiasing embeddings' or 'adversarial debiasing' could be explored during training to reduce inherent model biases.
    *   **Prompt Engineering**: Design prompts that explicitly encourage diversity and discourage stereotypes (e.g., 'Generate a story plot featuring a strong, independent protagonist from an underrepresented cultural background working as a scientist.').
    *   **Post-generation Filtering/Editing**: Implement a human-in-the-loop system where professional editors review and correct biased outputs. Automated tools for detecting problematic language or stereotypical representations can also support this process.

4.  **Evaluation Methods:**
    *   **Human Evaluation (Primary)**: Professional editors, writers, and literary critics from the publishing company would be the most important evaluators. They would assess:
        *   **Creativity & Originality**: Is the plot novel? Are characters engaging and unique?
        *   **Coherence & Consistency**: Does the plot make sense? Are character descriptions consistent with their roles and motivations?
        *   **Stylistic Fit**: Does the output align with the company's desired tone, genre, and quality standards?
        *   **Bias Detection**: Does the content exhibit harmful stereotypes or a lack of diversity?
        *   **Usability**: How effectively does the assistant aid human writers in their creative process?
    *   **Automated Metrics (Supportive)**:
        *   **Perplexity**: Measures how well the model predicts a sample of text; lower perplexity often indicates better language modeling.
        *   **Diversity Metrics**: Quantify the uniqueness and variety of generated outputs.
        *   **Readability Scores**: Assess the linguistic complexity and flow of the generated text.

**Real-world Challenges:**

1.  **Maintaining Creativity and Originality**: Generative models often interpolate from their training data. Ensuring truly novel, surprising, and non-derivative outputs, rather than mere variations of existing tropes, is a significant ongoing challenge.
2.  **Controlling 'Hallucinations'**: Models can generate factually incorrect, illogical, or inconsistent plot points or character traits, requiring careful human oversight.
3.  **Consistency Over Long Narratives**: Maintaining plot threads, character arcs, and world-building consistency across extensive story generations is exceptionally difficult for current LLMs.
4.  **Managing Subtle Bias & Stereotypes**: Despite mitigation efforts, subtle biases from vast, real-world training datasets can persist and manifest in outputs, demanding continuous monitoring and human review.
5.  **Ethical Concerns & Copyright**: Questions about authorship, originality, and potential unintentional replication of copyrighted material are paramount. Transparency regarding AI involvement is crucial.
6.  **Seamless Integration with Human Workflow**: The assistant should serve as a powerful co-creative tool, not a replacement. Designing an intuitive interface and workflow that empowers writers to easily iterate, refine, and steer AI-generated content is key.
7.  **Computational Resources**: Training and operating large-scale generative models demand substantial computational power and storage infrastructure.
8.  **Subjectivity of 'Good' Writing**: Defining and objectively measuring what constitutes a 'good' story or character is inherently subjective, making model optimization based on objective metrics complex.
9.  **Scalability**: Efficiently generating a high volume of diverse and high-quality content for multiple projects simultaneously presents a considerable scalability challenge.
